```python
# A1 = np.array((-1, 1))
# A2 = np.array((2, 5))
# A3 = np.array((5, -3))

A1 = np.array((-1, 2))
A2 = np.array((2, 3))
A3 = np.array((5, 4))

side_1 = np.sqrt(np.abs(A1[0] - A2[0]) ** 2 + np.abs(A1[1] - A2[1]) ** 2)
side_2 = np.sqrt(np.abs(A2[0] - A3[0]) ** 2 + np.abs(A2[1] - A3[1]) ** 2)
side_3 = np.sqrt(np.abs(A1[0] - A3[0]) ** 2 + np.abs(A1[1] - A3[1]) ** 2)

perimeters_expression = (side_1 + side_2 + side_3) * (side_2 + side_3 - side_1) * (side_1 + side_3 - side_2) * (side_1 + side_2 - side_3)
area = 0.25 * (perimeters_expression ** 0.5)
area
```
