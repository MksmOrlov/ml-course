Задача 3. Строки
Напишите функцию front_x(words), которая на вход принимает список строк и возвращает отсортированный по правилам:
1. слова, начинающиеся с символа "x", идут первыми
2. в лексикографическом порядке.

```python
def front_x(words: list[str]) -> list[str]:
    x_words = []
    empty_strings = []
    normal_words = []
    for word in words:
        if len(word) > 0 and word[0] == 'x':
            x_words.append(word)
        elif not word:
            empty_strings.append(word)
        else:
            normal_words.append(word)
    result = sorted(x_words)
    result.extend(empty_strings)
    result.extend(sorted(normal_words))
    return result
front_x(['mix', 'extra', '', 'x-files', 'xyz', 'xapple', 'apple'])
```

Задача 4. условия if-else. Напишите функцию is_prime(n), проверяющую простое ли число. Если число простое, функция должна вернуть True, иначе False (логическое значение, не строку!).

```python
def is_prime(n: int) -> bool:
    if not isinstance(n, int) or n < 1:
        raise ValueError("Число должно быть натуральным")
    if n < 4:
        return True
    for number in range(2, int(n**0.5) + 1):
        if n % number == 0:
            return False
    return True
print(is_prime(7))
print(is_prime(15))
print(is_prime(99))
print(is_prime(103))
```

Задача 5. срезы строк. Считайте строку.
Напишите функцию, которая принимает на вход строку, а возвращает 2 первых и 2 последних символа. Если длина исходной строки меньше 2 символов, необходимо вернуть пустую строку. Выведите на печать результат работы вашей функции, применённый к считанной строке.


```python
user_input = input("Введите строку: ")
def get_first_and_last_2(string: str) -> str:
    if len(string) <= 2:
        return ""
    first_2 = string[:2]
    last_2 = string[-2:]
    return first_2 + last_2
print(get_first_and_last_2(user_input))
```

Задача 6. Функции. Напишите функцию, для нахождения двойного факториала числа dfactorial(n)

```python
def dfactorial(n: int) -> int:
    if not isinstance(n, int):
        raise ValueError("Число должно быть неотрицательным и целым")
    stop = n % 2
    result = 1
    while n != stop:
        result *= n
        n -= 2
    return result
```

Задача 7. Представьте себя сотрудником деканата :(

```python
files_dir = r"files/task7"
sheet = "sheet1.txt"
mean = "mean1.txt"

EXAM = "экзамен"
AUTO = "автомат"
TYPES = (EXAM, AUTO)

expected_score = -1
calculated_score = -2
try:
    with open(f"{files_dir}/{mean}", mode='r') as average_score_file:
        expected_score = int(average_score_file.readline())
except FileNotFoundError:
    print(f"ERROR: file {files_dir}/{mean} not found")
sum_score = 0
count_score = 0
current_score = 0
try:
    with open(f"{files_dir}/{sheet}", mode='r', encoding="UTF-8") as scores:
        line = scores.readline()
        while line:
            if any(score_type in line for score_type in TYPES):
                current_score = int(line[line.find('(') - 2])
                count_score += 1
                sum_score += current_score
            line = scores.readline()
except FileNotFoundError:
    print(f"ERROR: file {files_dir}/{sheet} not found")
try:
    calculated_score = sum_score / count_score
except ZeroDivisionError:
    print("ERROR")
else:
    if expected_score == calculated_score:
        print("OK")
    else:
        print("ERROR")
```

Дополните лог в файле так, чтобы новое событие было записано вверху. Не забудьте указать порядковый номер события.


```python
EVENT = "event"
SEPARATOR = "-"
files_dir = r"files/task8"
log_filename = "log.txt"
file_path = f"{files_dir}/{log_filename}"

start_index = len(EVENT)
new_event = input("Введите новое событие")
try:
    with open(file_path, mode='r') as log_file:
        content = log_file.read()
        end_index = content.find(SEPARATOR) - 1
        last_number = int(content[start_index:end_index])
    with open(file_path, mode='w') as log_file:
        log_file.write(f"{EVENT} {last_number + 1} {SEPARATOR} '{new_event}'\n{content}")
except FileNotFoundError:
    print(f"File {file_path} not found, create new log file")
    with open(file_path, "w") as new_log_file:
        new_log_file.write(f"{EVENT} 1 {SEPARATOR} '{new_event}'")
```
