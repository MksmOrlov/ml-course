Задача 1. Часть 1.

```python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.linspace(-15, 15, 2000)
params = [(1, 1), (2, 1), (1, 2)]
colors = ['r', 'g', 'b']

fig, ax = plt.subplots(figsize=(10, 10))

for (a, b), color in zip(params, colors):
    y = []
    for x in x_values:
        xb = x ** b
        if abs(xb) < 0.05: # окрестность точки разрыва
            y.append(np.nan)
        else:
            y.append((xb + a ** b) / xb)
    ax.plot(x_values, y, color=color, label=f'a={a}, b={b}')

plt.ylim(-15, 15)
plt.legend(loc=3, fontsize=12)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title("$f(x) = \\frac{x^b+a^b}{x^b}$", fontsize=20)
plt.grid(True)
plt.show()
```
